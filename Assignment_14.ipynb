```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('pakistani_ecommerce_dataset.csv')
df.head()
df.isnull().sum()
```

```python
category_counts = df.groupby('product_category')['user_id'].count()
category_counts.sort_values(ascending=False).head(1)
```

```python
print(round(df['order_value'].mean(), 2))
```

```python
city_counts = df.groupby('city')['user_id'].count()
city_counts.sort_values(ascending=False).head(1)
```

```python
payment_counts = df.groupby('payment_method')['user_id'].count()
payment_counts.sort_values(ascending=False).head(1)
```

```python
avg_by_city = df.groupby('city')['order_value'].mean()
avg_by_city = avg_by_city.sort_values(ascending=False)
print(avg_by_city.round(2))

plt.figure(figsize=(8, 5))
plt.bar(avg_by_city.index, avg_by_city.values, color='green', width=0.5)
plt.xlabel('City')
plt.ylabel('Average Order Value (PKR)')
plt.title('Average Order Value by City')
plt.show()
```

## Dataset 2: Student Lifestyle & Study Habits

```python
df2 = pd.read_csv('student_lifestyle_dataset.csv')
```

```python
print(round(df2['study_hours'].mean(), 2))
```

```python
df2.groupby('gender')['sleep_hours'].mean().sort_values(ascending=False).round(2).head(1)
```

```python
corr = df2[['screen_time', 'sleep_hours']].corr().loc['screen_time', 'sleep_hours']
print(round(corr, 2))
```

```python
plt.hist(df2['daily_transport_cost'], bins=10, color='blue', edgecolor='black', alpha=0.7)
plt.xlabel('Daily Transport Cost (PKR)')
plt.ylabel('Number of Students')
plt.title('Histogram of Daily Transport Cost')
plt.show()
```
